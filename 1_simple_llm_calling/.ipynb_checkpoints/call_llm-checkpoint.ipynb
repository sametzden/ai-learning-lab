```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI



load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model = "gemma-4-31b-it")

response = llm.invoke("how many moons does Jupiter have?")
print(response.text)
```

```python
messages = [
    (
        "system",
        "You are a helpful assistant that translates English to French. Translate the user sentence.",
    ),
    ("human", "I love programming."),
]
response = llm.invoke(messages)
print(response.text)
```

```python
prompt = "write a single tagline for a quirky coffee shop. only output the tagline , nothing else"
llm = ChatGoogleGenerativeAI(model = "gemma-4-31b-it",temperature=0)

response = llm.invoke(prompt)
print(response.text)
```

```python
prompt = "write a single tagline for a quirky coffee shop. only output the tagline , nothing else"
llm = ChatGoogleGenerativeAI(model = "gemma-4-31b-it",temperature=1)
#higher temperature means more creativity
response = llm.invoke(prompt)
print(response.text)
```
